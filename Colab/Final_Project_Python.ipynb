# Bivariate Analysis → relation between two variables


# **Scatter Plot:**

```python

import seaborn as sns
plt.figure(figsize=(10, 6))
sns.scatterplot(x='ai_intensity_score', y='salary_usd', data=df)
plt.title('AI Intensity Score vs. Salary (USD)')
plt.xlabel('AI Intensity Score')
plt.ylabel('Salary (USD)')
plt.grid(True)
plt.show()
```

 # **What the Chart Shows:**

The **x‑axis** is AI Intensity Score (how strongly AI is used in a role or industry).

The **y‑axi**s is Salary (USD).

**Data points**: Each dot represents one job record in your dataset.

# **Features Used:**

1.AI Intensity Score → numeric variable .

2.Salary (USD) → numeric variable.

# **Explanation**:

1.The chart is exploring the relationship between AI intensity Score and salary levels.

2.The chart provides evidence for whether AI adoption is linked to better pay, which is a key insight for workforce and career planning.

3.If dots trend upward (higher AI intensity → higher salary), it suggests jobs requiring more AI skills tend to pay better.

4.If dots are scattered without a clear pattern, it means AI intensity doesn’t strongly affect salary.

# **Line Plot:**

```python
avg_ai_intensity_per_year = df.groupby('posting_year')['ai_intensity_score'].mean().reset_index()

plt.figure(figsize=(12, 6))
sns.lineplot(x='posting_year', y='ai_intensity_score', data=avg_ai_intensity_per_year, marker='o', color='blue')
plt.title('Average AI Intensity Score Over Years (2010-2025)')
plt.xlabel('Posting Year')
plt.ylabel('Average AI Intensity Score')
plt.grid(True)
plt.show()
```

**y-axis** represents the Average AI Intensity Score.

**Line Plot:** The line connects the average AI intensity score for each year, showing the trend over time.

1.posting_year (numeric variable).

2.ai_intensity_score (numeric variable).

# **Explanation:**

1.The line shows how AI Intensity Score has changed year by year.

2.The upward trend indicates that AI usage in jobs is steadily increasing from 2010 to 2025.

3.Each marker shows the average score for that year, making it easy to spot growth or dips.

4.It provides insights into the evolving landscape of job requirements and the growing importance of AI skills.

# **Correlation Heatmap:**

```python
# Select only numeric columns for correlation matrix
numeric_df = df.select_dtypes(include=['number'])

# Calculate the correlation matrix
corr_matrix = numeric_df.corr()

plt.figure(figsize=(12, 10))
sns.heatmap(corr_matrix, annot=True, cmap='coolwarm', fmt=".2f", linewidths=.5)
plt.title('Correlation Heatmap of Numerical Features')
plt.show()
```

# **What the Chart Shows:**

**Colors:** The color intensity and hue indicate the strength and direction of the correlation.
*   Warm colors (e.g., red) typically represent positive correlations.
*   Cool colors (e.g., blue) typically represent negative correlations.
*   Lighter colors indicate weaker correlations, while darker colors indicate stronger ones.

**Values:** The numbers displayed in each cell are the correlation coefficients, ranging from -1 (perfect negative correlation) to 1 (perfect positive correlation), with 0 indicating no linear correlation.

**Axes:**
* Both the x-axis and y-axis list the numerical features of the dataset.

# **Features Used:**

All numerical features in the dataset are used, including:
1.  posting_year
2.  ai_intensity_score
3.  salary_usd
4.  salary_change_vs_prev_year_percent
5.  automation_risk_score

# **Explanation:**

1.  This heatmap helps to quickly identify variables that are highly correlated with each other. For example, if ai_intensity_score shows a strong positive correlation with salary_usd, it reinforces the idea that jobs with higher AI intensity tend to offer better salaries.

2.  Conversely, a strong negative correlation (e.g., between automation_risk_score and salary_usd) might suggest that jobs with higher automation risk tend to have lower salaries.

3.  Understanding these relationships is crucial for building predictive models, as highly correlated features might indicate multicollinearity or provide insight into causal relationships that warrant further investigation.

4.  The diagonal elements are always 1, as a variable is perfectly correlated with itself.

# **histogram:**

```python
import seaborn as sns

plt.figure(figsize=(10,6))
sns.histplot(df['automation_risk_score'], bins=15, kde=True, color='red')
plt.title("Distribution of Automation Risk (%)")
plt.xlabel("Automation Risk (%)")
plt.ylabel("Frequency")
plt.grid(True)
plt.show()
```

# **What the Chart Shows:**

The chart displays the distribution of automation risk scores across all job records in your dataset.

1.The x‑axis represents Automation Risk (%) values (from low to high).

2.The y‑axis shows the frequency (how many jobs fall into each risk range).

3.The red bars represent counts, and the smooth KDE curve overlays the distribution shape.

# **Features Used:**

1.Numeric Variable (x-axis): automation_risk_score → continuous values between 0 and 1 (or 0-100%).

2.Frequency (y-axis): automatically calculated by Seaborn → number of records in each bin.

3.Bins: 15 intervals, grouping risk scores into ranges.

4.KDE Curve: shows the probability density (smooth trend of distribution).

5.Color: red bars for clarity.

# **Explanation:**

* The histogram reveals two peaks (clusters):

   * One around 0.3 → jobs with relatively low automation risk.

   * Another around 0.8 → jobs with high automation risk.

* This suggests a bimodal distribution: jobs tend to fall into either safer categories (less likely to be automated) or highly vulnerable categories (more likely to be automated).

* The KDE curve makes this clearer by showing the smooth probability density.

* Interpretation:

   * Industries or roles with low risk are more resilient to automation.

   * Roles with high risk may face significant disruption as AI adoption increases.

* This visualization helps identify which segments of the workforce are most exposed to automation threats.

# **boxplot**:

```python
import seaborn as sns
import matplotlib.pyplot as plt

plt.figure(figsize=(12, 7))
sns.boxplot(x="seniority_level", y="salary_usd", data=df)
plt.title("Salary Distribution by Seniority Level")
plt.xlabel('Seniority Level')
plt.ylabel('Salary (USD)')
plt.grid(True)
plt.show()
```

# **What the Chart Shows**:


* The chart compares salary distributions across different seniority levels (Intern, Junior, Mid, Senior, Lead, Executive).

* Each box represents the spread of salaries within that level:

  * The median (middle line inside the box).

  * The interquartile range (IQR) (the box itself, showing 25th-75th percentile).

  * Whiskers (lines extending from the box, showing the range of most salaries).

  * Outliers (individual points beyond whiskers, unusually high or low salaries).

# **Features Used**:

1.x-axis (Categorical): seniority_level → categories like Intern, Junior, Mid, Senior, Lead, Executive.

2.y-axis (Numeric): salary_usd → continuous salary values in USD.

3.Grid: Added for readability of salary ranges.

# **Explanation**:

1.Interns & Juniors: Lower median salaries, narrower boxes → salaries are more consistent and clustered.

2.Mid & Senior: Higher medians, wider boxes → salaries vary more, reflecting diverse compensation packages.

3.Lead & Executive: Highest medians, with longer whiskers and more outliers → compensation is less predictable, with some very high salaries.

4.Overall Insight: As seniority increases, both the median salary and the spread of salaries increase. This suggests higher roles not only pay more but also have greater variability (likely due to bonuses, stock options, or company differences).

# **Multivariate Analysis** → relation among 3+ variables

# **pivot tables:**

```python
pivot_table_salary = df.pivot_table(values='salary_usd', index='industry', columns='seniority_level', aggfunc='mean')
display(pivot_table_salary)
```

# **What the Chart Shows:**

*   **Rows:** Each row represents a specific industry.
*   **Columns:** Each column represents a seniority_level (Intern, Junior, Mid, Senior, Lead, Executive).
*   **Values:** The cells contain the average salary_usd for a given industry and seniority level combination.
*   This table provides a comprehensive view of how average salaries vary across different industries and how these salaries change as job roles become more senior within those industries.

# **Features Used:**

1.  industry (categorical variable, as index).
2.  seniority_level (categorical variable, as columns).
3.  salary_usd (numeric variable, as values, aggregated by mean).

**Explanation**:

* **Vertical insight** (within industry): For each industry, salaries rise with seniority. For example, IT shows a clear progression from Intern → Executive, with executives earning the highest averages.

* **Horizontal insight** (across industries): Comparing the same seniority level across industries reveals differences. For instance, Finance executives may earn more than Government executives, while Healthcare juniors may earn less than IT juniors.

* **Patterns:**

   * Interns → consistently lowest salaries across industries.

   * Executives → highest salaries, but with large variation depending on industry.

   * Mid/Senior levels → show steady growth, often reflecting industry competitiveness.

**Overall takeaway:** The pivot table highlights both career progression in salaries and industry-specific differences, making it a powerful tool for identifying where compensation is strongest or weakest.

# **grouped analysis:**

```python
grouped_analysis = df.groupby(['industry', 'seniority_level']).agg(
    avg_salary=('salary_usd', 'mean'),
    avg_ai_intensity=('ai_intensity_score', 'mean'),
    avg_automation_risk=('automation_risk_score', 'mean')
).reset_index()

display(grouped_analysis.head(10))
```

# **What the Analysis Shows:**

*   **Industry & Seniority Level:** The first two columns (`industry` and `seniority_level`) define each unique segment of the job market being analyzed.
*   **Average Salary (avg_salary):** Shows the mean salary in USD for each specific industry and seniority level combination.
*   **Average AI Intensity (avg_ai_intensity):** Indicates the average AI intensity score for jobs within that segment.
*   **Average Automation Risk (avg_automation_risk):** Represents the average automation risk score for jobs within that segment.

This table allows for a detailed comparison of these three critical metrics across various industry-seniority pairings.

# **Features Used:**

1.  industry (categorical variable for grouping)
2.  seniority_level (categorical variable for grouping)
3.  salary_usd (numeric variable, aggregated by mean)
4.  ai_intensity_score (numeric variable, aggregated by mean)
5.  automation_risk_score (numeric variable, aggregated by mean)

# **Explanation:**

*   **Comprehensive Overview:** This grouped analysis provides a granular view of how salaries, AI integration, and automation risks are interconnected within different professional contexts.
*   **Identifying Trends:**
    *   Whether higher AI intensity correlates with higher salaries within specific industries or seniority levels.
    *   How automation risk might decrease as seniority increases in certain sectors, or conversely, if some senior roles still carry high risk.
*   **Strategic Insights:** For workforce planning, this helps identify:
    *   Industries and seniority levels that are highly impacted by AI (high ai_intensity, potentially high or low automation_risk).
    *   Segments that might require targeted reskilling initiatives (high automation_risk).
    *   Career paths that offer promising salary growth in AI-driven environments.
*   **Comparing Segments:** This table enables direct comparisons between different industries for a given seniority level (e.g., 'Senior' in 'Information Technology' vs. 'Senior' in 'Education') or within an industry across seniority levels, offering a multi-faceted understanding of the job market landscape.
